```python
import numpy as np
```

```python
def fitness(u):
    return -u**2 + 5*u + 20
```

```python
positions = np.array([
    -9.6, -6.0, -2.6, -1.1, 0.6,
     2.3,  2.8,  6.3, 10.0
])
```

```python
n = len(positions)
velocities = np.zeros(n)

# PSO parameters
w = 0.5
c1 = 1.5
c2 = 1.5
```

```python
# Search boundaries
lower_bound = -10
upper_bound = 10
```

```python
max_iterations = 10
```

```python
pbest_positions = positions.copy()
pbest_fitness = fitness(positions)
```

```python
# Select the particle with the highest fitness
best_index = np.argmax(pbest_fitness)
gbest_position = pbest_positions[best_index]
gbest_fitness = pbest_fitness[best_index]
```

```python
print("INITIAL POPULATION")
print("-" * 65)
print("Particle | Position | Fitness")
for i in range(n):
    print(
        f"{i+1:8d} | {positions[i]:8.2f} | "
        f"{fitness(positions[i]):8.2f}"
    )

print("\nInitial Global Best Position:", gbest_position)
print("Initial Global Best Fitness:", gbest_fitness)
```

```python
# Perform PSO iterations
for iteration in range(max_iterations):

    # Random numbers for each particle
    r1 = np.random.rand(n)
    r2 = np.random.rand(n)

    # Update velocity
    velocities = (
        w * velocities
        + c1 * r1 * (pbest_positions - positions)
        + c2 * r2 * (gbest_position - positions)
    )

    # Update position
    positions = positions + velocities

    # Keep positions within search boundaries
    positions = np.clip(positions, lower_bound, upper_bound)

    # Calculate new fitness
    current_fitness = fitness(positions)

    # Update personal best
    improved = current_fitness > pbest_fitness
    pbest_positions[improved] = positions[improved]
    pbest_fitness[improved] = current_fitness[improved]

    # Update global best
    best_index = np.argmax(pbest_fitness)

    if pbest_fitness[best_index] > gbest_fitness:
        gbest_position = pbest_positions[best_index]
        gbest_fitness = pbest_fitness[best_index]

    # Display each iteration
    print(f"\nITERATION {iteration + 1}")
    print("-" * 65)
    print("Particle | Position | Velocity | Fitness")

    for i in range(n):
        print(
            f"{i+1:8d} | {positions[i]:8.4f} | "
            f"{velocities[i]:8.4f} | "
            f"{fitness(positions[i]):8.4f}"
        )

    print("Global Best Position:", round(gbest_position, 6))
    print("Global Best Fitness:", round(gbest_fitness, 6))
```

```python
print("\n" + "=" * 45)
print("FINAL RESULT")
print("=" * 45)
print("Best Position =", gbest_position)
print("Maximum Fitness =", gbest_fitness)

# Verify the mathematical maximum
exact_position = 2.5
exact_fitness = fitness(exact_position)

print("\nTHEORETICAL MAXIMUM")
print("Best Position =", exact_position)
print("Maximum Fitness =", exact_fitness)
```
